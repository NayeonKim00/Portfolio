## 금융상품의 구조

### 채권

- 채권이란 일정 기간 후 원금상환 및 이자지급조건이 발행시 확정되어 있는 차용증서를 의미합니다. 쉽게 말해 **돈을 빌려준 증거**라고 할 수 있습니다.

- 채권은 앞서 서술했듯 발행주체에 따라 국공채, 금융채, 회사채로 분류됩니다. 

- 채권은 이자 지급 형태에 따라 3가지로 구분이 가능합니다.
    1. 무이표채(Zero Coupon Bond) : 만기에 **액면가(face value)를 단 한번 지급**하는 형태를 가지고 있습니다.
    2. 이표채(Coupon Bond) : **만기까지 이자(Coupon)을 지급**하고, 만기일에 액면가와 이자를 지급하는 형태를 가지고 있습니다.
    3. 주택담보대출채권(Mortgage) : 만기까지 이자(Coupon)을 지급하나, face value 지급은 없는 형태를 가지고 있습니다.
    
- 이외에도 만기가 정해지지 않은 경우(ex : 연금)나 물가와 연동되어 이자율이 변동하는 채권(물가연동채권) 등이 존재합니다.

#### 채권의 가격결정론

##### Yield To Maturity

- 만기 수익률(Yield to Maturity)이란 채권을 만기까지 보유하였을때 얻을 수 있는 수익률을 의미합니다. 이때, 모든 수익은 만기수익률(YTM)과 동일한 수익률로 재투자되어야만 합니다. 

- 이는 채권의 **미래 현금흐름의 현재가치의 총합과 채권가격을 같게 해 주는 할인률**이라고 할 수 있습니다.

- 국채와 같이 신뢰도가 높은 채권의 만기수익률은 작은 경향을 보이며, 회사채와 같이 비교적 신뢰가 떨어지는 채권의 만기수익률은 큰 경향을 보입니다.

- 만기수익률과 채권가격은 **역의 관계**를 갖습니다. 즉, 만기수익률이 커지면 채권가격은 하락하고 만기수익률이 작아지면 채권가격은 상승합니다.

##### 채권가격결정방법

- 할인채의 경우, 액면가가 $F$이고 만기수익률이 $i$이며 투자 기간이 $T$년인 할인채라면 이 할인채의 현재 가격은 다음과 같이 결정됩니다.

$$ P=\frac{F}{(1+i)^T} $$


- 이표채의 경우, 액면가가 $F$이고 만기수익률이 $i$이며 투자 기간이 $T$년이고 동일한 이자가 $C$로 주어질 때 이 이표채의 가격은 다음과 같이 결정됩니다.

$$ P=\sum_{t=1}^{T} \frac{C}{(1+i)^t}+\frac{F}{(1+i)^T} $$


##### 채권의 Duration

채권의 Duration이란 이자율의 변화율에 대한 채권가격의 변화율, 즉 이자율에 대한 채권가격의 탄력성을 의미합니다. 식으로는 다음과 같습니다.

$$ D=-\frac{ \frac{ dP }{ P } } { \frac{ dR }{ R } } = -\frac{dP}{P}\frac{R}{dR}$$


정의에 의해 duration을 계산해보면 다음과 같은 결과를 얻게 됩니다.

$$ D=\frac{ \sum_{t=1}^{T} PV(CF_t) \times \tau_t } {P } = \frac{ \sum_{t=1}^{T} \frac{ \tau_t \times C} {(R)^t} +  \frac{F}{(1+i)^T} \times \tau_T}{P}$$

이 듀레이션을 Macaulay duration이라고 부릅니다. 여기서
    
1. $PV(CF_t)$ : 현금흐름의 현재가치
2. $\tau_t$ : 시간 t에서의 시간의 길이를 연도 단위로 표기한 것
3. $C$ : 이자
4. $T$ : 만기일까지 남은 기간 의미(배당을 하는 기간을 의미한다)
5. $F$ : 만기시 지급하는 액면가
6. $P$ : 채권가격
7. $R$ : 이자율($1+\text{만기수익률}$ 을 의미한다)
    
위의 Macaulay duration에서 R을 나눈 값, 즉 $$ \frac{D}{R}=-{\frac{dP}{P}}\frac{1}{dR}$$을 수정 듀레이션(Modified duration)이라고 합니다

##### 채권의 Convexity

채권가격은 비선형이기 때문에, 이자율의 변화에 따른 채권가격의 변화를 추적할 때 선형근사인 Duration만을 사용하는 것은 위 그림과 같이 큰 오차를 야기합니다. 따라서 이를 보정하기 위해 2차 근사항까지 사용하게 되는데, 이를 Convexity(볼록성)이라고 부릅니다. 볼록성을 수식으로 정의하면 다음과 같습니다.

$$ \text{Convexity} \equiv \frac{1}{f(r)} \frac{d^2f(r)}{{dr}^2} $$

따라서 2차 테일러 근사식

$$ \frac{f(r+\Delta r)-f(r)}{f(r)} = \frac{f'(r)}{f(r)} \Delta r + \frac{f''(r) }{2}\frac{{\Delta r}^2}{f(r)} $$

에서 이자율이 변할 때 채권가격의 변화율은 다음과 같이 결정됨을 알 수 있습니다.   

$$ \frac{f(r+\Delta r)-f(r)}{f(r)} = \frac{\Delta f(r)}{f(r)}=-M \times \Delta r + \frac{C}{2}\times {\Delta r}^2$$

여기서 $M$은 수정 듀레이션을, $C$는 볼록성을 뜻합니다

## 주식

- 주식은 주식회사에 출자를 한 주주가 주주로서 회사에 대해 가지는 법률상의 권리와 의무의 단위입니다. 간단하게 **주식은 회사에 대한 소유권**이라고 이해할 수 있습니다. 또한 주식은 **잉여자본에 대한 청구권**으로도 이해할 수 있습니다.

- 주식을 보유한 주주는 유한 책임(Limited liability)을 가집니다. 즉, 주주는 자신이 출자한 양 이상의 책임을 지지 않습니다.

### 주식의 가격결정론

#### 배당할인모형(Dividend Discount Model, DDM)

주식을 보유할 때 오는 손익은 배당과 주식의 가치 변동에서 오게 됩니다. 더 정확히 표현하자면 시점 t에서 주식에 1기동안 투자했을때 기대 수익률$1+R$은 $$1+R=\frac{E_t[P_{t+1}]+E_t[D_{t+1}]}{P_t}$$로 주어집니다.


이제 기대 수익률 R이 시간에 대해 불변하고 expectation $E_t$를 편의를 위해 빼보겠습니다. 위 식을 변형하면 $$P_t=\frac{P_{t+1}+D_{t+1}}{1+R}$$가 성립합니다. 그리고 이 식은 다음 기에도 성립해야 하기 때문에 $$P_{t+1}=\frac{P_{t+2}+D_{t+2}}{1+R}$$ 또한 성립해야 합니다. 따라서 $$P_t=\frac{P_{t+1}+D_{t+1}}{1+R}= \frac{\frac{P_{t+2}+D_{t+2}}{1+R}+D_{t+1}}{1+R} = \frac{P_{t+2}}{(1+R)^2}+\frac{D_{t+2}}{(1+R)^2}+\frac{D_{t+1}}{(1+R)}$$이 성립하게 됩니다. 그리고 이 다음 기에도 동일한 식이 성립하므로 $$P_{t+2}=\frac{P_{t+3}+D_{t+3}}{1+R}$$이 성립하고, 이를 기존의 $P_t$ 식에 대입해 정리하면 $$P_t= \frac{P_{t+3}}{(1+R)^3}+\frac{D_{t+3}}{(1+R)^3}+\frac{D_{t+2}}{(1+R)^2}+\frac{D_{t+1}}{(1+R)}$$이 성립하게 됩니다. 이를 T 기간 동안에 대해 구해보면 $$P_t=\sum_{i=1}^{T}\frac{D_{t+i}}{(1+R)^i} + \frac{P_{t+T}}{(1+R)^T}$$라는 결과를 얻게 되고 이를 무한히 반복하게 되면 $$P_t=\sum_{i=1}^{\infty}\frac{D_{t+i}}{(1+R)^i}$$가 성립하게 됩니다. 이를 배당할인모형이라 부릅니다.

#### Gordon의 성장 모델

Gordon의 성장 모델은 위 배당할인모형을 기본으로 합니다. 여기에 기존에는 $\{D_n\}_{n \in [t,T]}$였던 배당이 $D_{t+1}=(1+G)D_t$라는 일정한 증가율을 가진다는 가정을 더한 결과입니다.


따라서 위 배당할인모형의 결론이었던$$P_t=\sum_{i=1}^{\infty}\frac{D_{t+i}}{(1+R)^i}$$을 이용하면 $$P_{t}=\sum_{i=1}^{\infty}\frac{(1+G)^{i-1}}{(1+R)^{i}}D \ (단\ D는 \ 초기배당)$$라는 결론을 얻을 수 있습니다. 이를 정리하면 $$P_{t}=\frac{D}{1+R} \sum_{i=1}^{\infty}\frac{ (1+G)^{i-1}}{(1+R)^{i-1}} =\frac{D}{1+R} \sum_{i=0}^{\infty}\Bigg(\frac{1+G}{1+R}\Bigg)^i$$이라는 결론을 얻을 수 있습니다.

이를 무한등비급수의 합 공식을 이용해 정리하면 $$P_{t}=\frac{D_0(1+G)}{R-G} $$라는 결과를 얻게 됩니다.

여기서 배당의 장기 성장률 $G$를 결정하는 요소는 유보비율($b$)과 자기자본이익률($\text{ROE}$)의 곱인 $b \times \text{ROE}$ (지속가능성장률)가 됩니다. 그 이유는 기업이 창출한 순이익 중 배당으로 지급하고 남은 유보금을 다시 자본으로 환입하고, 기존의 부채비율과 투자 수익률(ROE)을 일정하게 유지하면서 재투자한다고 가정하기 때문입니다.

## 파생상품 기초

### 파생상품의 정의

- 자본시장법에 따르면 파생상품이란 **원본대비 손실률이 100% 초과 가능성이 있는 상품**을 의미합니다
- 경제학적으로 파생상품이란 **특정 지수나 현물 가격과 같은 기초자산(underlying asset)의 변동에 의거해 자산의 가치가 결정되는 상품**을 의미합니다.

### 파생상품의 분류

일반적으로, 파생상품은 크게 3가지로 분류합니다.
1. 선도(Forward) : 기초자산이나 기초자산의 가격 등에 의해 산출된 금전 등을 **장래의 특정시점에 인도할 것을 약정**하는 계약입니다.
2. 옵션(Option) : 당사자의 어느 한쪽의 의사표시에 의하여 기초자산의 가격 등에 의해 산출된 금전 등을 수수하는 거래를 성립시킬 수 있는 **권리를 부여하는 계약**입니다.
3. 스왑(Swap) : 장래의 일정기간동안 미리 정한 가격으로 기초자산의 가격 등에 의해 산출된 금액을 **교환할 것을 약정하는 계약**입니다.

## 선물(Futures)

- 선물은 **미래의 일정한 시점(만기일)에 일정한 양(거래단위)과 질을 갖춘 특정 상품(기초자산)을 미리 정한 가격(선물가격)으로 매매**하기로 **현재 시점에서 약속한 계약**을 의미합니다. 선물 계약에는 아래 4가지 요소는 미리 결정되어 있으며, 시장에서 선물 가격을 결정하게 됩니다.
    1. 기초자산 : 선물 거래시 인수도 대상이 되는 자산을 말합니다. 상품이 기초자산이 되기 위해서는 시장유동성, 거래조건 표준화, 시장가격의 자율 경쟁을 통한 결정, 높은 가격변동성, 시장정보의 자유롭고 공개적인 취득 가능성 등의 조건이 필요합니다.
    2. 거래단위 : 계약단위를 의미하며, 선물 1 계약의 크기를 뜻합니다. 이 거래단위가 너무 크다면 유동성이 부족해져 선물시장규모가 축소될 우려가 있고, 거래단위가 너무 작다면 유동성은 확대되나 선물시장의 변동성이 커질 위험이 존재합니다.
    3. 인수도 장소 : 실물자산의 경우 중요한 요소로 선물 만기시에 기초자산을 인수도할 장소를 의미합니다.
    4. 만기월 : 현물이 인도됨으로써 계약의무가 이행되는 날짜를 말합니다.

### 선물의 손익 구조
미래 시점에서 기초자산 인수도가 일어나는 만큼, 선물의 손익 구조는 미래 시점의 기초자산의 가격에 의존합니다.

#### 선물 매수자
선물 매수자의 경우, 미래 시점의 기초자산 가격이 비싸지면 기초자산을 싸게 구매한 것이기 때문에 이익을 보며, 미래 시점의 기초자산 가격이 싸지면 기초자산을 비싸게 구매한 것이기 때문에 손해를 봅니다. 따라서 선물 매수자의 Payoff는 다음과 같이 주어집니다.
$$\Pi_T=(S_T-F_t)X$$
이때, 밑첨자 $T$는 만기월, $S_T$는 만기 시 기초자산의 가격, $F_t$는 선물가격, X는 선물 거래량을 의미합니다.

#### 선물 매도자
선물 매도자의 경우, 미래 시점의 기초자산 가격이 비싸지면 기초자산을 싸게 판매한 것이기 때문에 손해를 보며, 미래 시점의 기초자산 가격이 싸지면 기초자산을 비싸게 판매한 것이기 때문에 이익을 봅니다. 따라서 선물 매도자의 Payoff는 다음과 같이 주어집니다.
$$\Pi_T=(F_t-S_T)X$$
이때, 밑첨자 $T$는 만기월, $S_T$는 만기 시 기초자산의 가격, $F_t$는 선물가격, X는 선물 거래량을 의미합니다.

매도자와 매수자의 payoff를 동시에 표현하기 위해 다음과 같이 서술합니다.
$$\Pi_T=(F_t-S_T)X \ 
\begin{cases}
if X>0, & X \ 계약 \ 매도, \newline
if X<0, & X \ 계약 \ 매수,
\end{cases}$$
즉, $X>0$이면 $X$계약 매도한 것이며, $X<0$이면 $X$계약 매수한 것입니다.

## 선물가격과 현물가격

일반적으로 선물가격과 현물가격은 같지 않습니다. 그 이유는 선물 가격은 **미래의 재화**를 기준으로 하는 만큼, 현물을 보유하는 과정에서의 손익이 포함되어 결정되기 때문입니다.

선물 가격 $F_t$는 다음과 같이 결정됩니다. $$F_t=S_t+보유비용-보유수익$$ 이러한 이론을 보유비용모형이라고 부릅니다.

당연하겠지만, 일반적으로 $F_t \neq S_t$입니다. 그리고 $$\begin{cases}
F_t > S_t, & Contango \ Market, \newline
F_t < S_t,  & Backwardation \ Market,
\end{cases} $$ 이라고 부릅니다.

또한, Basis $B_t$를 $$B_t \equiv F_t-S_t$$로 정의하는데, $$\begin{cases}
B_t>0, & Contango \ Market, \newline
B_t<0,  & Backwardation \ Market,
\end{cases} $$ 이라고 할 수 있습니다.

## 옵션(Option)

옵션 또한 미래 재화를 다룬다는 점에서는 선물과 비슷한 면을 가집니다. 하지만 선물과 옵션의 가장 큰 차이점은 선물은 손해를 보던 이익을 보던 계약에 따른 의무를 매도자와 매수자 모두 이행해야 하지만, 옵션의 경우 행사를 할지 말 지 결정할 **권리**를 거래한다는 점입니다.

옵션에는 두가지 종류가 있습니다.
1. Call option : **살** 권리를 의미합니다.
2. Put option : **팔** 권리를 의미합니다.

또한 옵션의 중도 행사 가능 여부에 따라 두가지로 나뉩니다.
1. American option : 옵션매입자는 **중도 권리 행사가 가능**합니다.
2. European option : 옵션매입자는 **만기에만 권리 행사가 가능**합니다.

그리고 옵션에서는 기초자산의 매도와 매수를 결정하는 것이 아닌, **권리**의 매도와 매수를 구분합니다. 예를 들어, 석유 call option **매수자**의 경우 석유를 **살 권리를 구매**한 것이며, 석유 call option **매도자**는 석유를 **살 권리를 판매**한 것을 의미합니다. 석유 put option의 경우는 put option **매수자**는 석유를 **팔 권리를 매수**한 것이며, 석유 put option **매도자**는 석유를 **팔 권리를 매도**한 것이 됩니다. 이를 다르게 표기하면 옵션매수자는 **매매이행(계약이행)을 청구할 권리**를 가지며, 옵션매도자는 **매매이행(계약이행)을 할 의무**만을 가지게 됩니다.

### 옵션의 손익구조

#### Call option의 손익구조

call option은 **살 권리**에 해당합니다. 따라서 call option 매수자는 미래 현물가격이 행사가격보다 비쌀때 살 권리를 행사할 것이며, 미래 현물가격이 행사가격보다 쌀 때 살 권리를 행사하진 않을 것입니다. 왜냐하면 옵션을 행사하지 않고 현물시장에서 사는게 더 싸기 때문입니다.

이를 정리하면 $$\begin{cases}
if \ K\leq S_T, & 권리를\ 행사한다 \newline
if \ K>S_T,  & 권리를\ 행사하지\ 않는다.
\end{cases} $$

그리고 권리 행사를 통해 얻는 이익은
$$\begin{cases}
if \ K\leq S_T, & \Pi_T=S_T-K \newline
if \ K>S_T,  & \Pi_T=0
\end{cases} $$
이 나옵니다. 이를 정리하면 $$\Pi_T=Max[S_T-K,0]$$ 이라는 사실을 알 수 있습니다.
따라서 Call option의 손익구조는 $$\Pi_T=(c_0-c_T)X \; \begin{cases}
if \ X>0, & X\ 만큼\ 매도 \newline
if \ X<0,  & X\ 만큼\ 매수.
\end{cases} \newline where \ c_T=Max[S_T-K,0], \; c_0는\ 매수시\ 옵션프리미엄$$
라고 할 수 있습니다.

#### Put option의 손익구조

put option은 **팔 권리**에 해당합니다. 따라서 put option 매수자는 미래 현물가격이 행사가격보다 쌀때 팔 권리를 행사할 것이며, 미래 현물가격이 행사가격보다 비쌀 때 팔 권리를 행사하진 않을 것입니다. 왜냐하면 옵션을 행사하지 않고 현물시장에 파는게 더 비싸기 때문입니다.

이를 정리하면 $$\begin{cases}
if \ K\geq S_T, & 권리를\ 행사한다 \newline
if \ K<S_T,  & 권리를\ 행사하지\ 않는다.
\end{cases} $$

그리고 권리 행사를 통해 얻는 이익은
$$\begin{cases}
if \ K\geq S_T, & \Pi_T=K-S_T \newline
if \ K<S_T,  & \Pi_T=0
\end{cases} $$
이 나옵니다. 이를 정리하면 $$\Pi_T=Max[K-S_T,0]$$ 이라는 사실을 알 수 있습니다.
따라서 Put option의 손익구조는 
$$\Pi_T=(p_0-p_T)X \; \begin{cases}
if \ X>0, & X\ 만큼\ 매도 \newline
if \ X<0,  & X\ 만큼\ 매수
\end{cases} \newline where \ p_T=Max[K-S_T,0], \; p_0는\ 매수시\ 옵션프리미엄$$
라고 할 수 있습니다.

### Put-Call Parity

한 만기와 기초자산, 그리고 행사가격을 갖는 put가격과 call 가격 중 하나라도 알게 된다면, 나머지 하나의 가격은 바로 추론이 가능합니다. 이는 put-call parity 때문인데, **put-call parity는 european option이라는 가정**만 필요합니다. 기초자산의 가격 분포나 이론적 옵션가격, 기초자산의 분산 등 다른 요소에 대한 가정이 전혀 필요없기  때문에 option 차익거래 과정에서 매우 중요한 요소로 꼽힙니다. 

put-call parity는 $$C^e-P^e=S_0-Ke^{-rT}$$로 주어집니다. 여기서
1. $C^e$ : call option premium
2. $P^e$ : put option premium
3. $S_0$ : 기초자산의 현재가
4. $K$ : 행사가격
5. $r$ : 이자율
6. $T$ : 만기까지 남은 시간

입니다.

### Binomial Tree Model

이항 모델(binomial tree model)는 현재 주가에서 미래 주가로의 움직임이 **일정 비율로 상승하거나 하락하는 2가지 경우만을 가진다**고 가정하는 모델입니다.

이항모델을 이용해 옵션의 가격을 계산할 때 도입되는 개념이 바로 델타 헤징(delta hedging)입니다. 이는 포트폴리오 구성 시 미래 기초자산의 가치에 상관없이 포트폴리오 가치가 일정하게 유지하게 해 주는 델타를 찾는 문제라고도 할 수 있습니다.

좀 더 구체적으로 예시를 들어보겠습니다. 현재 기초자산의 가격이 20, 보유기간은 3개월(즉, one step의 크기가 3개월), 연단위 연속복리 이자율이 12%이고 미래 가능한 두가지 자산가격이 $1.1S_0=22\ $ or $\ 0.9S_0=18\ $로 주어진다고 해 봅시다. 그리고 우리는 $\Delta$만큼의 기초자산을 구매하고 행사가격이 $K=21$인 1계약의 call을 매도하는 포트폴리오를 구축한다고 해보겠습니다. 이때, $\Delta$를 잘 조정한다면 미래 포트폴리오의 가치는 기초자산의 가격이 어떻게 되든 상관없이 일정하게 만들 수 있을 것입니다. 이 $\Delta$를 찾는 것을 델타 헤징(delta hedging)이라고 부릅니다.

각 두가지 경우에 대한 미래 포트폴리오 가치는 다음과 같습니다.

<div align="center">

| $S_T=22$ | $S_T=18$ |
|:---:|:---:|
| $22\Delta - \max[S_T-K, 0] = 22\Delta - 1$ | $18\Delta - \max[S_T-K,0] = 18\Delta$ |

</div>

미래 포트폴리오 가치가 $S_T$에 상관없이 항상 일정하기 위해서는 $22\Delta-1=18\Delta$이면 됩니다.
이를 만족하는 $\Delta=0.25$임을 알 수 있습니다.
다음으로 우리는 이 포트폴리오를 구성하는 비용이 $- 20\Delta +c^e$임을 알고 있습니다. 이때, 미래에 포트폴리오 수익은 항상 일정하므로 차익거래가 일어나지 않기 위해서는 **현재 포트폴리오 payoff+만기 포트폴리오 payoff=0**을 만족해야 합니다. 왜냐하면 이 포트폴리오는 아무런 위험을 가지고 있지 않기 때문입니다.($S_T$가 어떻게 변하든 이 포트폴리오의 수익은 항상 일정함을 기억합시다) 따라서 $$-20\cdot0.25+c^e+(22\cdot0.25-1)e^{-0.12\cdot \frac{1}{4}}=0$$을 만족하는 옵션 가격 $c^e$는 $$-5+c^e+4.5e^{-0.03}=0,\ c^e=0.635$$라는 결론을 얻을 수 있습니다.

이를 일반화해보면 다음과 같이 나옵니다. 주가가 미래 오르는 비율을 $u$, 주가가 미래 떨어지는 비율을 $d$라고 하면 미래 주가는 $Su$, $Sd$로 나타납니다. 주가가 올랐을때의 옵션 가치를 $f_u$, 주가가 떨어졌을때의 옵션 가치를 $f_d$라고 해 봅시다. 동일하게 우리는 $\Delta$만큼의 기초자산을 구매하고 1계약의 call을 매도하는 포트폴리오를 구축한다고 해보겠습니다.(행사가격은 만기 옵션 가치를 결정짓는데 이미 사용되었습니다.) 그러면 두가지 경우에 대한 미래 포트폴리오 가치는 다음과 같이 결정됩니다.

<div align="center">

|$S_T=Su$|$S_T=Sd$|
|:---:|:---:|
|$Su\Delta -f_u$|$Sd\Delta - f_d$|

</div>

여기서 미래 포트폴리오 가격을 $S_T$에 상관없이 일정하게 해 주는 $\Delta$는 $$Su\Delta -f_u=Sd\Delta - f_d,\ \Delta=\frac{f_u-f_d}{Su-Sd}$$가 됨을 알 수 있습니다.
그리고 이를 바탕으로 옵션의 현재가치 $f$를 계산해보면 $$f-S\Delta+(Su\Delta-f_u)e^{-rT}=0,\ f=S\Delta-(Su\Delta-f_u)e^{-rT}$$라는 결론을 얻을 수 있습니다.

위의 $$f=S\Delta-(Sd\Delta-f_d)e^{-rT}$$에 $$\Delta=\frac{f_u-f_d}{Su-Sd}$$를 넣어 정리해보면

$$
\begin{aligned}
f &= S\Delta - (Sd\Delta - f_d)e^{rT} \\[1.5em]
  &= e^{-rT}(S\Delta e^{rT} - Sd\Delta + f_d) \\[1.5em]
  &= e^{-rT}[S\Delta(e^{rT}-d) + f_d] \\[1.5em]
  &= e^{-rT}\left[S\frac{f_u-f_d}{Su-Sd}(e^{rT}-d) + f_d\right] \\[1.5em]
  &= e^{-rT}\left[\frac{f_u-f_d}{u-d}(e^{rT}-d) + f_d\right] \\[1.5em]
  &= e^{-rT}\left[\frac{e^{rT}-d}{u-d}f_u + \left(1-\frac{e^{rT}-d}{u-d}\right)f_d\right]
\end{aligned}
$$

의 꼴을 갖게 됩니다. 여기서 $p\equiv\frac{e^{rT}-d}{u-d}$라고 정의하면 $$f=e^{-rT}[pf_u+(1-p)f_d]$$꼴이 됩니다. 즉, $p$가 마치 가격이 올라갈 확률처럼 행동하게 되며, 옵션의 가격이 미래 옵션의 기대가격의 현재가치꼴이 됩니다.

$ud=1$이라는 가정 하에 $u$와 $d$는 $$u=e^{\sigma \sqrt{\Delta t}}, \ d=e^{-\sigma \sqrt{\Delta t}}$$로 계산할 수 있습니다.

### 블랙-숄즈 모형(Black–Scholes model)

옵션 가격 결정 모델로 가장 유명한 모델인 블랙-숄즈 모형은 가격의 분포가 로그 정규분포(Log-normal distribution) 라는 가정에서 시작하는 옵션 가격결정모형입니다. 

European call option에 대한 블랙 숄즈 옵션가격은 다음과 같이 계산됩니다.

$$ c^{e}  = S_t \cdot N(d1) - K \cdot e^{-r \cdot T} \cdot N(d2)$$

European put option에 대한 블랙 숄즈 옵션가격은 다음과 같이 계산됩니다.


$$p^{e} = K \cdot e^{-r \cdot T} \cdot N(-d2) - S_t \cdot N(-d1)$$

여기서

- $ c^{e} $는 이론적 European call option의 가격입니다.
- $ p^{e} $는 이론적 European put option의 가격입니다.
- $ S_t $ 는 현재 기초자산의 가격입니다.
- $ K $ 는 옵션의 행사가격입니다.
- $ r $ 는 무위험 이자율입니다.
- $ T $ 는 연 단위로 표기한 만기까지 남은 기간입니다.
- $ N $ 는 표준정규분포의 누적확률분포함수를 의미합니다.
- $ d1 $와 $ d2 $는 다음과 같이 결정됩니다.

$$d1 = \frac{ ln(\frac{S_t}{K}) + (r + \frac{σ^{2}}{2}) \cdot T}{ σ\cdot \sqrt{T}} \newline d2 = d1 - σ \cdot \sqrt{T}$$

여기서 $ σ $는 기초자산의 변동성입니다.

또한, $N(-d2)$는 european put option이 만기에 행사될 확률을 의미하며 $N(d2)$는 만기에서 european call option이 행사될 확률을 의미합니다.

### 파생상품의 리스크 지표: 민감도(Greeks) 분석

블랙-숄즈 모형을 통해 옵션의 '가격'을 구했다면, 리스크 관리자(Risk Manager)나 트레이더의 다음 질문은 "기초자산의 가격이나 변동성이 변할 때, 내 포트폴리오의 가치는 얼마나 변하는가"로 넘어갑니다. 이를 측정하는 지표들을 그릭스(Greeks)라고 부릅니다.

특히 헷징(Hedging) 전략에서 가장 핵심이 되는 두 가지 지표는 델타($\Delta$)와 감마($\Gamma$)입니다.

#### 1. 델타 (Delta, $\Delta$): 방향성 리스크 (Directional Risk)
델타는 기초자산 가격($S$) 변화에 따른 옵션 가격($c$ 또는 $p$)의 변화율을 의미합니다. 수학적으로는 옵션 가격을 기초자산 가격으로 편미분한 1차 도함수입니다.

$$ \Delta = \frac{\partial f}{\partial S} $$

- **블랙-숄즈 모형에서의 델타:**
  - Call Option의 델타: $\Delta_c = N(d_1)$ (항상 0에서 1 사이의 값을 가짐)
  - Put Option의 델타: $\Delta_p = N(d_1) - 1$ (항상 -1에서 0 사이의 값을 가짐)

- **리스크 관리 관점의 해석 (Delta Hedging):** 
  델타는 단순히 가격 변화율을 넘어 **'헤지 비율(Hedge Ratio)'**을 의미합니다. 만약 내가 델타가 0.5인 콜옵션을 1계약(통상 기초자산 100주) 매도했다면, 나의 포트폴리오는 기초자산 가격 상승 시 손실을 보는 위험에 노출됩니다. 이를 완벽하게 방어(Delta Neutral)하려면, 기초자산 $0.5 \times 100 = 50$주를 현물 시장에서 매수하면 됩니다. 이것이 프론트 오피스의 가장 기본적인 델타 헤징 전략입니다.

#### 2. 감마 (Gamma, $\Gamma$): 볼록성 리스크 (Convexity Risk)
감마는 기초자산 가격($S$) 변화에 따른 델타($\Delta$)의 변화율을 의미합니다. 수학적으로는 옵션 가격을 기초자산 가격으로 두 번 편미분한 2차 도함수입니다. 채권에서 배웠던 **'볼록성(Convexity)'의 파생상품 버전**이라고 이해하면 정확합니다.

$$ \Gamma = \frac{\partial \Delta}{\partial S} = \frac{\partial^2 f}{\partial S^2} $$

- **블랙-숄즈 모형에서의 감마 (Call, Put 동일):**
  $$ \Gamma = \frac{N'(d_1)}{S \sigma \sqrt{T}} $$
  (단, $N'(x)$는 표준정규분포의 확률밀도함수)

- **리스크 관리 관점의 해석 (Gamma Risk & Gap Risk):**
  델타 헤징의 가장 큰 약점은 기초자산 가격이 조금만 변해도 델타 값 자체가 변해버린다는 점입니다(비선형성). 기초자산 가격이 서서히 움직일 때는 델타 헤징만으로 방어가 가능하지만, 밤사이 악재가 터져 주가가 폭락하는 등 **'점프(Jump)'**가 발생하면 델타만으로는 방어막이 뚫려버립니다. 감마가 높은 포트폴리오일수록 이러한 급변동 장세에서 예상치 못한 거대한 손실(또는 이익)을 보게 됩니다. 
  
  따라서 고도화된 리스크 관리를 위해서는 델타 중립(Delta Neutral)뿐만 아니라, 다른 옵션 계약을 섞어 감마 중립(Gamma Neutral)까지 맞추는 전략이 필요합니다.

#### 테일러 전개(Taylor Expansion)로 보는 시장 리스크
우리가 배운 델타와 감마는 결국 기초자산 가격 변화($\Delta S$)에 따른 포트폴리오 가치 변화($\Delta f$)를 테일러 전개로 근사한 것입니다.

$$ \Delta f \approx \Delta \times (\Delta S) + \frac{1}{2} \Gamma \times (\Delta S)^2 $$

이는 앞서 채권 파트에서 이자율 변화($\Delta r$)에 따른 채권 가격 변화율을 **수정 듀레이션(1차)과 볼록성(2차)**으로 근사했던 것과 수학적으로 완전히 동일한 철학을 공유하고 있습니다.

# 신용 리스크

신용 위험은 간단히 서술하면 거래 상대방이 합의된 조건에 따른 의무를 부분적 혹은 전체적으로 수행하지 못할 가능성으로 정의된다. 이는 매출채권 등에도 적용되는 risk로, 거의 모든 회사가 노출되어 있는 리스크라고 할 수 있다. 금융시장에서의 신용 리스크는 대출, 지급보증, 그리고 파생상품 거래 등 신용이 공여되는 모든 금융거래에 광범위하게 적용된다.

신용 리스크는 특히 금융권에서 매우 큰 의미를 갖는데, 금융권에서 신용을 기반으로 한 거래가 많이 발생하기 때문이다. (대출 등을 다루는 은행에서 크게 부각된다)

신용 리스크의 관리가 중요한 이유 또한 시장 리스크와 매우 유사하다. 신용 리스크가 높은 부실 여신(대출 및 채권)이 증가하면 결국 전체 자산 포트폴리오의 가치가 하락하여 기업이 단기 채무에 대응하지 못하는 유동성 위기에 빠질 가능성이 높아지기 때문이다. 또한 예측 가능한 신용 리스크는 리스크 프리미엄 형태로 대출 상품의 금리에 더해지는 만큼, 적절한 신용 리스크 예측은 매우 중요하다. 따라서 금융기관은 통계적으로 예측 가능한 '예상 손실(EL)'에 대해서는 적절한 대손충당금을 적립하여 대비하고, 이를 초과하여 발생하는 극단적인 '예상외 손실(UL)'에 대해서는 충분한 자기자본(Capital)을 보유함으로써 대규모 자산 부실이 유동성 위기나 은행의 파산으로 전이되는 것을 막아야만 한다.

신용 리스크는 2가지 접근 방법을 갖는다.

1. Credit risk in a strict sense : 채무자가 default하고 원금지불능력을 상실하는 위험만을 고려한다. 즉 미래 발생 가능한 경우는 default or not default 두 가지 경우만 존재한다.

2. Credit risk in a broader sense : 채무자가 default 할 확률 자체를 리스크로 판단하며, 이때의 결과값은 [0,1] 사이의 연속된 값으로 표현된다.

Credit risk in a strict sence는 매우 많은 채무자로 이뤄진 Portfolio(ex : 은행 개인 대출 portfolio)에서 중요하며, Credit risk in a broader sence는 포트폴리오에 매우 큰 비중의 채무자가 소수 존재하는 경우 많이 쓰인다.

일반적으로 신용 위험에는 세 가지 정의상의 특징이 존재한다.

1. 노출 : 노출이란 채무불이행이나 채무이행 능력에 불리한 변화를 겪을 수 있는 포트폴리오 금액을 의미한다.

2. 우도 : 우도는 채무자가 의무를 불이행할 확률을 의미한다.

3. 회수율 : 채무불이행이 발생하는 경우 회수할 수 있는 양을 의미한다.

IRB(Internal Ratings-Based Approach), 내부등급법은 바젤은행감독위원회가 정하고 있는 최소요건을 충족하여 감독당국의 승인을 받은 은행이 자체 추정한 리스크 측정요소를 활용하여 신용리스크에 대한 위험가중자산을 산출하는 방법을 의미한다. 이 내부 평가의 주요 변수는 다음과 같다.

$$예상 손실(Expected \ Loss) = EAD *  LGD  *  PD$$

여기서 

- 부도 확률 (PD, Probability of Default): 거래 상대방이 정해진 기간(보통 1년) 내에 부도를 낼 확률.

- 부도 시 손실률 (LGD, Loss Given Default): 부도가 발생했을 때, 담보 회수 등을 거치고도 최종적으로 떼이는 금액의 비율.

- 부도 시 익스포저 (EAD, Exposure At Default): 부도가 발생하는 시점에 거래 상대방에게 빌려준 총 금액(위험 노출액).

을 의미한다. 우리는 예상 손실에 대비하기 위해 해당 손실을 상품의 금리에 반영하여 리스크 프리미엄 형태로 흡수하고, 이에 대해 미리 대손충당금을 쌓아 대비한다.

예상 손실과 다른 개념으로, 예상외 손실 (Unexpected Loss, UL)이 존재한다. IMF 외환위기나 서브프라임 모기지 사태처럼, 통계적 평균을 뛰어넘어 꼬리(Tail) 부분에서 발생하는 거대한 실제 손실을 의미한다. 이 손실은 '유동성 위기나 파산'을 불러오는 진짜 주범으로 자기자본(Regulatory Capital)을 충분히 보유함으로써 흡수해야 한다.

## 구조적 모형(Structural Model) 기반 PD(Probability of Default) 추정

앞에서는 머신러닝 모델을 이용하여 PD를 추정하는 방법을 알아보았다. 이번에는 수리통계적 프레임워크를 기반으로 하는 구조적 모델에 대해 알아보자.

### 1. 머튼 모형 (Merton Model, 1974)

머튼 모형은 "기업의 주식(Equity)은 기업의 총자산(Asset)을 기초자산으로 하고, 부채(Debt)를 행사가격으로 하는 콜옵션(Call Option)과 동일하다"는 아이디어를 기반으로 전개된다. Call option은 살 권리로, 만일 기초자산 가격이 행사가격보다 높아진다면 그 권리를 행사하게 된다. 이를 기반으로 머튼 모형을 이해해보면 다음과 같은 통찰이 나온다. 주주는 기업의 총자산이 부채보다 더 크다면(잉여자본금이 존재한다면) 그에 대한 청구권을 행사한다(Call option의 행사). 하지만 기업의 총자산이 부채보다 작다면(음의 자본) 주주는 청구권을 행사하지 않을 것이다. 이것이 바로 유한책임의 핵심이다. 따라서 주식의 가치는 $V_E = \max(V_A - D, 0)$ 가 된다. 이는 정확하게 Call option의 가치와 동일하다. (단, $V_E$는 주식 가치, $V_A$ 는 자산 가치, $D$는 기업 부채)

더 자세하게 서술하기 위해 다음 식을 보자. 기업의 자산 가치 $V_t$가 기하브라운 운동을 따른다고 가정한다. 부채는 만기 $T$에 상환하는 제로쿠폰 부채 $D$로 단순화한다.

**Default 조건** : $ V_T < D $ 즉 만기에 기업의 자산 가치가 부채보다 적다

주주는 자산을 콜옵션으로 보유하고 있으므로 Black-Scholes 공식을 적용한다.

$$ V_{E} = V_A N(d_1) - D e^{-r(T-t)} N(d_2) $$

$$ d_1 = \frac{\ln(V_A / D) + (r + \frac{1}{2}\sigma_A^2)(T-t)}{\sigma_A \sqrt{T-t}} $$
$$ d_2 = d_1 - \sigma_A \sqrt{T-t} $$

여기서 $V_E$는 주식의 시장가치, $V_A$는 기업 총자산 가치를 의미한다.

**부도확률(PD)**:

우리가 알고 있는 사실은, Call option이 행사될 확률은 $N(d_2)$이다. 그렇다면 Call option이 행사되지 않을 확률은 $N(-d_2)$이다. (이는 put이 권리행사될 확률과 동일하다). 앞에서 Call option이 행사되지 않는 것은 부도가 난 상황과 동일하다는 것을 배웠다. 따라서 다음과 같은 부도 확률을 얻게 된다.

$$ PD = N(-d_2) $$

이를 계산하기 위해서는 우리는 $V_A$, $\sigma_A$를 계산할 수 있어야 한다. 따라서 ito's Lemma를 도입한다. 여기서 알아둬야 할 사실은 콜옵션의 델타($\Delta = \frac{\partial V_E}{\partial V_A}$)는 블랙-숄즈 공식에서 $N(d_1)$이라는 사실이다. 이를 변동성에 적용하면 다음과 같은 두 번째 방정식이 유도된다. 

$$\sigma_E V_E = \frac{\partial V_E}{\partial V_A} \sigma_A V_A = N(d_1) \sigma_A V_A$$

따라서 

$$\sigma_A = \frac{V_E \sigma_E}{V_A N(d_1)}  $$

가 성립한다

이렇게 우리는 PD를 추정하기 위한 준비를 마쳤다. 하지만 우리가 얻은 것은 비선형 연립방정식이므로, 계산이 매우 까다롭다. 따라서 이 상황에서는 최적화 알고리즘을 이용해 근사적으로 계산을 수행한다.

하지만 이러한 머튼 모델도 다양한 한계점을 보인다

1. 비현실적인 가정: 기업의 자산 가치가 기하브라운운동(GBM)을 따른다고 가정하며, 주식 수익률이 정규분포를 따른다는 가정은 현실의 두꺼운 꼬리(Fat Tail) 현상을 반영하지 못한다.

2. 만기 전 파산 무시: 오직 만기 시점에만 채무 불이행 여부를 판단하므로, 그 이전에 발생하는 유동성 위기나 파산을 포착하지 못한다.

3. 정보의 불완전성: 기업 자산의 시장 가치와 변동성을 정확히 측정하기 어렵고, 실제 파산 구조는 모델보다 복잡하다.

4. 시장 거래 비용 및 세금: 모델은 거래 비용, 세금, 부채 재조정 비용 등을 고려하지 않는다. 

#### 참고 : 이토의 보조정리를 이용한 증명 과정

총자산($V_A$)과 주식($V_E$)의 기하브라운운동(GBM) 가정먼저 기업의 총자산과 주식 가치가 각각 다음의 확률미분방정식을 따른다고 가정하자.

자산 가치의 변화: $dV_A = \mu_A V_A dt + \sigma_A V_A dW$

주식 가치의 변화: $dV_E = \mu_E V_E dt + \sigma_E V_E dW \quad$

주식 가치 $V_E$는 자산 가치 $V_A$와 시간 $t$의 함수인 $V_E(V_A, t)$이다. 여기에 이토의 보조정리를 적용하여 $dV_E$를 전개하면 다음과 같은 긴 식이 나온다. $$dV_E = \left( \frac{\partial V_E}{\partial t} + \mu_A V_A \frac{\partial V_E}{\partial V_A} + \frac{1}{2} \sigma_A^2 V_A^2 \frac{\partial^2 V_E}{\partial V_A^2} \right) dt + \left( \frac{\partial V_E}{\partial V_A} \sigma_A V_A \right) dW \quad $$

브라운 운동 항($dW$)의 계수 매칭(식 1)과 (식 2)는 결국 같은 주식 가치의 변화($dV_E$)를 나타낸다. 따라서 무작위한 충격을 나타내는 $dW$ 앞의 계수(확산항)는 서로 완벽하게 일치해야 한다. $$\sigma_E V_E = \frac{\partial V_E}{\partial V_A} \sigma_A V_A$$여기서 주식($V_E$)을 자산($V_A$)에 대한 콜옵션으로 볼 때, 기초자산의 변화에 따른 옵션 가격의 변화율 $\frac{\partial V_E}{\partial V_A}$는 바로 콜옵션의 델타($\Delta$)인 $N(d_1)$이 된다. 이를 대입하면 최종 수식이 완벽하게 증명된다. $$\sigma_E V_E = N(d_1) \sigma_A V_A$$

```python
from scipy.stats import norm
import numpy as np

def merton_pd(V, D, r, sigma_A, T):
    d2 = (np.log(V/D) + (r - 0.5*sigma_A**2)*T) / (sigma_A * np.sqrt(T))
    return norm.cdf(-d2)

# 예시: 자산가치 100, 부채 80, r=0.05, sigma_A=0.3, T=1
print(merton_pd(100, 80, 0.05, 0.3, 1))   # 약 0.048 (4.8%)
```

### 2. KMV 모형 (KMV-Merton 모형)

머튼 모형은 다양한 한계점을 가지고 있다. 머튼 모델을 실제 시장 데이터로 확장한 것이 KMV 모형이다. KMV 모델은 기업의 자산 가치가 총부채 아래로 떨어진다고 무조건 부도가 나는 건 아니라는 사실에 주목했다. 따라서 다음과 같은 부도 거리 개념을 도입하였다.

Distance-to-Default (DD):

$$DD = \frac{V_A - DP}{V_A \sigma_A}$$

여기서 $V_A$는 자산의 시장 가치, DP는 부도점, $\sigma_A$는 자산의 변동성을 의미한다. 또한 KMV 모델은 부도점(DP)에 대해서도 역사적 데이터를 기반으로 다음과 같은 결론을 내놓았다. 

$$DP = \text{단기부채 (STD, 1년 이내 만기)} + 0.5 \times \text{장기부채 (LTD)}$$

만일 자산가치의 분포가 로그 정규분포를 가정한다면 다음과 같이 식이 변하게 된다

$$ DD = \frac{\ln(V_A / DP) + (\mu - \frac{1}{2}\sigma_V^2)T}{\sigma_V \sqrt{T}}$$

이를 기반으로 예상 부도 확률(Expected Default Frequency, EDF)은 다음과 같이 계산한다.

$$EDF = N(-DD)$$

하지만 실제로는 금융 데이터가 정규분포를 따르지 않는 경우가 많다. 이를 해결하기 위해 실제 현업에서는 과거 역사적 부도 데이터베이스를 기반으로 계산된 DD와 대응되는 EDF를 찾는 식으로 부도 확률을 추정한다.

또한 머튼 모델과 KMV 모델 다 상장된 대기업에만 적용이 가능하다. 시장에서 거래되는 주식 가격(이 가격이 주식의 시장가를 나타내게 된다) 등이 존재해야만 모델이 작동하기 때문이다.

### 예상외 손실(UL)과 바젤 K-function
앞에서는 예상 손실을 모델링하는 방법에 대해 배웠다. 하지만 매우 중요하면서도 기업의 존립을 위태롭게 하는 변수는 예상 외 손실(Unexpected Loss, UL)이다. 이 예상 외 손실은 말 그대로 예상이 어렵고, 변동성이 커서 자기자본으로 대응해야 한다.

바젤위원회 IRB(내부등급법)에서 요구하는 자기자본 요구량(K)은 아래 K-function으로 계산한다. K는 Unexpected Loss를 커버하기 위한 위험가중자산(RWA) 비율을 의미하며, 은행은 K에 익스포저인 EAD를 곱한 만큼을 자기자본을 보유해야 한다.

가장 중요한 K는 다음과 같이 계산한다.

$$K = \left[ LGD \times N\left( \frac{G(PD) + \sqrt{R} \cdot G(0.999)}{\sqrt{1-R}} \right) - LGD \times PD \right] \times MA$$

여기서

$G(\cdot) = N^{-1}(\cdot)$ : 표준정규 누적분포 역함수

R : 자산과 거시경제 충격 간 상관계수 (기업 규모에 따라 다름. 대기업일수록 높으며 중소기업일수록 낮음)

$MA = \frac{1+(M-2.5)b}{1-1.5b}$ : Maturity Adjustment

b = $[0.11852 - 0.05478 \ln(PD)]^2$ 

로 주어진다


```python
from scipy.stats import norm

def basel_k(pd, lgd=0.45, r=0.12, m=2.5):
    g_pd = norm.ppf(pd)
    g_999 = norm.ppf(0.999)
    ma = (1 + (m - 2.5) * 0.11852 - 0.05478 * np.log(pd)) / (1 - 1.5 * (0.11852 - 0.05478 * np.log(pd))**2)
    term = lgd * norm.cdf( (g_pd + np.sqrt(r) * g_999) / np.sqrt(1 - r) ) - lgd * pd  
    return term * ma

print(f"PD=1%일 때 K ≈ {basel_k(0.01):.4f} (자기자본 8% 기준 위험가중자산 비율)")
```
